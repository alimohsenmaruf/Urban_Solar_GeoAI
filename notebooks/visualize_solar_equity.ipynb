```python
from pathlib import Path
import geopandas as gpd

BASE_DIR = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
SOLAR_PATH = BASE_DIR / "outputs" / "vectors" / "solar_rooftop_potential.geojson"
EQUITY_PATH = BASE_DIR / "outputs" / "vectors" / "canopy_equity_assessment.geojson"

solar_gdf = gpd.read_file(SOLAR_PATH)
equity_gdf = gpd.read_file(EQUITY_PATH)

print(f"Total candidate roofs for solar: {len(solar_gdf)}")
print(f"Est. total generation capacity: {solar_gdf['est_kwh_per_year'].sum():,.1f} kWh/year")
solar_gdf.head()
```

```python
from pathlib import Path
import leafmap

BASE_DIR = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
RASTER_PATH = str(BASE_DIR / "data" / "raw" / "urban_solar_sample.tif")

m = leafmap.Map(center=[30.2650, -97.7415], zoom=18)

# Add Satellite Base Layer
m.add_raster(RASTER_PATH, layer_name="Aerial Satellite Imagery")

# Overlay Solar Rooftop Layer
m.add_gdf(
    solar_gdf,
    layer_name="Solar Rooftop Potential (kWh/yr)",
    style={"color": "#ffaa00", "weight": 2, "fillOpacity": 0.4}
)

m
```
